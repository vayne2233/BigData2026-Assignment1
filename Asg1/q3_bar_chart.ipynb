```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# x-axis
datasets = ['GSM8K', 'MATH']

acc_top = [0.158, 0.038]
acc_25 = [0.124, 0.032]
acc_50 = [0.121, 0.020]
acc_bottom = [0.114, 0.028]

x = np.arange(len(datasets))
width = 0.15

fig, axes = plt.subplots(figsize=(8, 5), dpi=300)

# plot 8 bar
rects1 = axes.bar(x - 1.5 * width, acc_top, width, label='Top', color='green')
rects2 = axes.bar(x - 0.5 * width, acc_25, width, label='25%', color='darkblue')
rects3 = axes.bar(x + 0.5 * width, acc_50, width, label='50%', color='orange')
rects4 = axes.bar(x + 1.5 * width, acc_bottom, width, label='Bottom', color='royalblue')

axes.set_title('Accuracy', loc='left', fontsize=14, pad=25)
axes.set_xticks(x)
axes.set_xticklabels(datasets, fontsize=12)

for rects in [rects1, rects2, rects3, rects4]:
  axes.bar_label(rects, fmt='%.3f', padding=3, fontsize=9)

# add horizontal line
axes.yaxis.grid(True, linestyle='-', alpha=0.5)
axes.set_axisbelow(True)

# 圖例置頂置中
axes.legend(
    loc='upper center',
    bbox_to_anchor=(0.5, 1.12),
    ncol=4,
    frameon=False,
    fontsize=12,
)

plt.tight_layout()
plt.savefig('q3_bar_chart.png', dpi=300)
plt.show()
```
